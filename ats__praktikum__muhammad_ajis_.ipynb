```python
"""
PEMANTAUAN BEBAN HARIAN PEMBANGKIT LISTRIK
Nama  : [MUHAMMAD AJIS]
NIM   : [4232501014]
Kelas : RPE Pagi A

MASALAH:
    Operator pembangkit perlu memantau beban listrik (kW) tiap jam selama
    24 jam untuk:
      1. Mengetahui jam yang beban-nya NORMAL, WARNING, atau OVERLOAD.
      2. Menghitung total energi harian (kWh), beban rata-rata, beban
         puncak, dan load factor.
      3. Memperkirakan biaya energi harian.

CARA KERJA:
    1. Pengguna memilih data contoh atau input manual.
    2. Data 24 jam disimpan dalam list `beban`. Pada input manual,
       try-except memastikan data berupa angka dan tidak negatif
       (diulang dengan while sampai benar).
    3. Perulangan for memeriksa tiap jam, lalu if-elif-else menentukan
       status: NORMAL, WARNING (>= 80% kapasitas), OVERLOAD (> 100%).
    4. Statistik harian dihitung dan ringkasan ditampilkan.
"""

# ---------------- PARAMETER SISTEM ----------------
KAPASITAS_KW = 500.0        # kapasitas pembangkit (kW)
BATAS_WARNING = 0.80        # WARNING jika beban >= 80% kapasitas
TARIF_PER_KWH = 1444.70     # tarif listrik (Rp/kWh)
JUMLAH_JAM = 24

# Data contoh beban per jam (kW), pukul 00:00 sampai 23:00
DATA_CONTOH = [
    180, 170, 165, 160, 175, 210,   # 00 - 05
    280, 350, 390, 360, 340, 330,   # 06 - 11
    345, 355, 370, 395, 430, 480,   # 12 - 17
    520, 490, 420, 330, 260, 200,   # 18 - 23
]


def baca_beban_manual():
    """Membaca beban 24 jam dari keyboard dengan validasi try-except."""
    data = []
    jam = 0
    while jam < JUMLAH_JAM:
        teks = input(f"Beban jam {jam:02d}:00 (kW): ")
        try:
            nilai = float(teks)
            if nilai < 0:
                raise ValueError("beban tidak boleh negatif")
        except ValueError as err:
            print(f"  Input tidak valid ({err}). Coba lagi.")
            continue            # ulangi jam yang sama
        data.append(nilai)      # simpan ke list
        jam += 1
    return data


def tentukan_status(nilai_kw):
    """Mengembalikan status operasi berdasarkan persentase kapasitas."""
    persen = nilai_kw / KAPASITAS_KW
    if persen > 1.0:
        return "OVERLOAD"
    elif persen >= BATAS_WARNING:
        return "WARNING"
    else:
        return "NORMAL"


def main():
    print("=" * 50)
    print("Nama  : [MUHAMMAD AJIS]")
    print("NIM   : [4232501014]")
    print("Kelas : RPE Pagi A")
    print("=" * 50)
    print("\n=== PEMANTAUAN BEBAN HARIAN PEMBANGKIT ===")
    print(f"Kapasitas: {KAPASITAS_KW:.0f} kW | "
          f"Batas warning: {BATAS_WARNING * 100:.0f}%\n")

    # --- 1. Pilih sumber data ---
    pilihan = input("Pilih data: [1] Data contoh  [2] Input manual : ").strip()
    if pilihan == "2":
        beban = baca_beban_manual()
    else:
        beban = DATA_CONTOH.copy()

    # --- 2. Pemeriksaan tiap jam ---
    print("\nJam    Beban (kW)   % Kapasitas   Status")
    print("-" * 46)

    jumlah_status = {"NORMAL": 0, "WARNING": 0, "OVERLOAD": 0}
    jam_bermasalah = []     # list jam yang WARNING / OVERLOAD

    for jam, nilai in enumerate(beban):
        status = tentukan_status(nilai)
        persen = nilai / KAPASITAS_KW * 100
        print(f"{jam:02d}:00  {nilai:9.1f}   {persen:10.1f}%   {status}")

        jumlah_status[status] += 1
        if status != "NORMAL":
            jam_bermasalah.append(jam)

    # --- 3. Statistik harian ---
    total_energi = sum(beban) * 1          # kWh (kW x 1 jam)
    rata_rata = total_energi / len(beban)  # kW
    puncak = max(beban)
    jam_puncak = beban.index(puncak)
    load_factor = rata_rata / puncak * 100 if puncak > 0 else 0
    biaya = total_energi * TARIF_PER_KWH

    # --- 4. Ringkasan ---
    print("\n=== RINGKASAN HARIAN ===")
    print(f"Total energi     : {total_energi:,.1f} kWh")
    print(f"Beban rata-rata  : {rata_rata:.2f} kW")
    print(f"Beban puncak     : {puncak:.1f} kW pada pukul {jam_puncak:02d}:00")
    print(f"Load factor      : {load_factor:.2f}%")
    print(f"Perkiraan biaya  : Rp {biaya:,.0f}")
    print(f"Jumlah jam -> NORMAL: {jumlah_status['NORMAL']}, "
          f"WARNING: {jumlah_status['WARNING']}, "
          f"OVERLOAD: {jumlah_status['OVERLOAD']}")

    if jumlah_status["OVERLOAD"] > 0:
        print("\nKESIMPULAN: Terjadi OVERLOAD. Perlu penambahan kapasitas "
              "atau pengaturan (load shedding) pada jam: "
              + ", ".join(f"{j:02d}:00" for j in jam_bermasalah))
    elif jumlah_status["WARNING"] > 0:
        print("\nKESIMPULAN: Beban mendekati batas pada jam: "
              + ", ".join(f"{j:02d}:00" for j in jam_bermasalah))
    else:
        print("\nKESIMPULAN: Seluruh beban dalam kondisi aman.")


if __name__ == "__main__":
    main()
```
